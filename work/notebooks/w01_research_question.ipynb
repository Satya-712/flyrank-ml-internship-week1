## Quick Look at the Data

The starter dataset contains 30,000 pages. The correlation between search volume and impressions is approximately 0.001, showing that search volume alone is not enough to explain page visibility. The median word count of declining pages is around 2909 words, suggesting that content length alone is not the main factor behind declining performance. These observations support the need for a machine learning approach to identify pages that require content refresh.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
import pandas as pd

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

print("Total Pages:", len(df))

corr = df["search_volume"].corr(df["impressions_90d"])
print("Correlation (Search Volume vs Impressions):", round(corr, 3))

median_wc = df[df["trend_direction"] == "down"]["word_count"].median()
print("Median Word Count (Declining Pages):", round(median_wc))
```
